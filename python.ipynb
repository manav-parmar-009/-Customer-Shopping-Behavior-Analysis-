```python
import matplotlib.pyplot as plt
```

```python
#Customers by Loyalty Tier
g1=df.groupby("loyalty_tier")["customer_id"].nunique()
plt.bar(g1.index,g1.values)
plt.title("Customers by Loyalty Tier")
plt.xlabel("Loyalty Tier")
plt.ylabel("number of customers")
```

```python
#Age Distribution
plt.hist(df["age"],bins=10,color="red",edgecolor="black")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
```

```python
#Income Distribution
plt.hist(df["income_k"],bins=10,edgecolor="black",color="yellow")
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
```

```python
#box plot of income
plt.boxplot(df["income_k"])
plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
```

```python
#Customers by Experiment Group
g2 = df.groupby("experiment_group")["customer_id"].nunique()

plt.pie(g2.values, labels=g2.index, autopct="%1.1f%%")
plt.title("Customers by Experiment Group")
plt.show()
```

```python
#weekly purchases trend
g1 = df.groupby("week")["purchased"].sum()

plt.plot(g1.index, g1.values, marker="o")
plt.xlabel("Week")
plt.ylabel("Total Purchases")
plt.title("Purchases by Week")
plt.show()
```
